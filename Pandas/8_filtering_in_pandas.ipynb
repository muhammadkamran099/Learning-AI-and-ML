```python
import numpy as np
import pandas as pd
```

```python
salary = np.array([45000, 52000, 0, 61000, 48000])
department = np.array(["developer", "marketing",
"intern", "sales", "support"])
salary_series = pd.Series(salary,
index=department,
name="Salary")
```

```python
salary_series > 50000
```

```python
salary_series.loc[salary_series > 50000]
```

```python
mask = (salary_series >= 40000) & (salary_series < 60000)
salary_series.loc[mask]
```

```python
print(salary_series.loc[(salary_series > 45000) & (salary_series < 60000)])
```
