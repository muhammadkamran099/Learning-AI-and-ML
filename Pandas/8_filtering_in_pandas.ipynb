```python
import numpy as np
import pandas as pd
```

```python
salary = np.array([45000, 52000, 0, 61000, 48000])
department = np.array(["developer", "marketing",
"intern", "sales", "support"])
salary_series = pd.Series(salary,
index=department,
name="Salary")
```

```python
salary_series > 50000
```

```python
salary_series.loc[salary_series > 50000]
```

```python
mask = (salary_series >= 40000) & (salary_series < 60000)
salary_series.loc[mask]
```

```python
print(salary_series.loc[(salary_series > 45000) & (salary_series < 60000)])
```

```python
print(salary_series.loc[(salary_series > 30000) | (salary_series < 40000)])
```

```python
print(salary_series.loc[(salary_series == 0)])
```

```python
# it gives all salary values that is not zero
print(salary_series.loc[~(salary_series == 0)])
```

```python
print(salary_series.loc[salary_series.between(40000, 60000)])
```

```python
print(salary_series.loc[salary_series.index.isin(["developer", "sales"])])
```

```python
print(salary_series.loc[salary_series.isin([0, 45000, 61000])])
```

```python
mask = salary_series > 50000
print(mask)
print(mask.sum())
print(mask.any())
print(mask.all())
print(len(salary_series.loc[mask]))
```

```python
salary_series.loc[salary_series <= 50000] == 50000
salary_series
```
