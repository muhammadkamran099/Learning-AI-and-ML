```python
import numpy as np
import pandas as pd
```

```python
salary_values = np.array([25000, 29000, 31000, 0, 41000, 56000, 38000])
salary_series = pd.Series(salary_values, name="salary")
salary_series
```

```python
# Temporarily converts the values to floats without modifying the original Series.
salary_series.astype(float)
```

```python
# No change in original series
salary_series
```

```python
salary_series.astype(str)
```

```python
salary_series.astype("string")
```

```python
salary_series.astype(bool)
```

```python
# Permanently converts the Series values to string data type by assigning the result back to the original variable.
salary_series = salary_series.astype("string")
```

```python
salary_series
```

```python

temp_array = np.array([29.5, 31.0,
28.7, 32.2, 30.4, 27.9])
temp_series = pd.Series(temp_array, name ='Temperature')

temp_series.mean()
```

```python
temp_array = np.array([29.5, 31.0,
28.7, 32.2, 30.4, 27.9])
temp_series = pd.Series(temp_array, name ='Temperature')
temp_series = temp_series.astype(int)

temp_series.mean()
```
