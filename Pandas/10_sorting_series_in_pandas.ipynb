```python
import numpy as np
import pandas as pd
```

```python
salary = np.array([45000, 52000, 0, 61000, 48000])
department = np.array(["developer", "marketing",
"intern", "sales", "support"])
salary_series = pd.Series(salary,
index=department,
name="Salary")
```

```python
# sort by value -> low to high
salary_series.sort_values()
```

```python
# sort by value -> high to low
salary_series.sort_values(ascending=False)
```

```python
# Sort by index - A to Z
salary_series.sort_index() 
```

```python
# Sort by index - Z to A
salary_series.sort_index(ascending=False)
```

```python
# sorting returns a new series, the original one stays unchanged
sorted_series = salary_series.sort_values()
print(salary_series)
print(sorted_series)
```

```python
salary_series.nlargest(2)
```

```python
salary_series.nsmallest(2)
```

```python
print(salary_series)
print("--------------")
print(salary_series.rank())
```

```python
# Inplace change the original series instead of returning of new one
salary_series.sort_values(inplace=True)
salary_series
```
