```python
import numpy as np
import pandas as pd
```

```python
billing = pd.Series([450.75, 1200.50, 980.00, 1550.25, 760.40, 2300.00, 890.10, 1475.60, 650.00, 3100.75], name="billing")
billing
```

```python
def give_discount(bill: float):
    if bill > 1000:
        bill = bill - ( bill * 10 / 100)
        return round(bill, 2)
    return bill
```

```python
billing.apply(give_discount)
```
