```python
import numpy as np
import pandas as pd
```

```python
departments_dup = ["developer", "developer", "hr", "sales", "sales", "support"]
salaries = [45000, 52000, 0, 61000, 58000, 72000]
salary_dup = pd.Series(salaries, index=departments_dup,
name="Salary")
salary_dup
```

```python
salary_dup.index.is_unique
```

```python
# resetting the index
# without drop=True (old index becomes a column)
salary_dup.reset_index()
```

```python
# With drop=True (old index is discarded)
salary_dup.reset_index(drop=True)
```
