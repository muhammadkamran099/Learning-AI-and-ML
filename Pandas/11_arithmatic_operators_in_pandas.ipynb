```python
import numpy as np
import pandas as pd
```

```python
salary = np.array([45000, 52000, 0, 61000, 48000])
department = np.array(["developer", "marketing",
"intern", "sales", "support"])
salary_series = pd.Series(salary,
index=department,
name="Salary")
salary_series
```

```python
print(salary_series - 5)
print("-----------------------")
print(salary_series.sub(5))
```

```python
print(salary_series + 5)
print("-----------------------")
print(salary_series.add(5))
```

```python
print(salary_series * 5)
print("-----------------------")
print(salary_series.mul(5))
```

```python
print(salary_series / 5)
print("-----------------------")
print(salary_series.div(5))
```

```python
print(salary_series // 5)
print("-----------------------")
print(salary_series.floordiv(5))
```

```python
print(salary_series % 2)
print("-----------------------")
print(salary_series.mod(2))
```
