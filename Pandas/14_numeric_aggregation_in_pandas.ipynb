```python
import pandas as pd
```

```python
departments = [ "engineering", "marketing", "sales", "hr", "finance", "support", "design", "operations" ]

salary = [ 52.0, 47.5, 61.0, float("nan"), 58.5, 43.0, 49.5, 54.0 ]

salary_series = pd.Series(salary, index=departments, name="salary")

salary_series

```

```python
# Count only umber of non-missing values
salary_series.count()
```

```python
# Sum of all non-missing values

salary_series.sum()
```

```python
# Average of all non-missing values 
salary_series.mean()
```

```python
# Middle value when sorted (ignores NaN)
salary_series.median()
```

```python
# Smallest value (ignores NaN)
salary_series.min()
```

```python
# Largest value (ignores NaN)
salary_series.max()
```
