```python
import pandas as pd
import numpy as np
```

```python
department = [ "engineering", "marketing", "sales", "hr", "finance", "support", "design", "operations" ]

salary_series = pd.Series([52.0, 47.5, 61.0, np.nan, 58.5, np.nan, 49.5, 54.0 ], index=department, name="salary")

salary_series

```

```python
# Count only number of non-missing values
salary_series.count()
```

```python
# Count Nan values
salary_series.isna().sum()
```

```python
# Size count every row including missing values(Nan)
salary_series.size
```

```python
# Sum of all non-missing values

salary_series.sum()
```

```python
# Average of all non-missing values 
salary_series.mean()
```

```python
# skipna = False stop pandas ignoring Nan
salary_series.mean(skipna = False)
```

```python
# Middle value when sorted (ignores NaN)
salary_series.median()
```

```python
# Smallest value (ignores NaN)
salary_series.min()
```

```python
# Largest value (ignores NaN)
salary_series.max()
```
