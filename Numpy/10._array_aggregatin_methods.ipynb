```python
import numpy as np
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

sales.sum()
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

sales.sum(axis=0)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])
# Coumns wise sum
sales.sum(axis=0)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

# Rows wise sum
sales.sum(axis=1)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])


sales.mean()
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

# column wise mean
sales.mean(axis=0)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

# column wise mean
sales.mean(axis=1)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

sales.min(axis=1)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

sales.max()
```
