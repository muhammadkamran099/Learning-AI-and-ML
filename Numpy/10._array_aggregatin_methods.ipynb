```python
import numpy as np
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

sales.sum()
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

sales.sum(axis=0)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])
# Coumns wise sum
sales.sum(axis=0)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

# Rows wise sum
sales.sum(axis=1)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])


sales.mean()
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

# column wise mean
sales.mean(axis=0)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

# column wise mean
sales.mean(axis=1)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

sales.min(axis=1)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

sales.max()
```

```python
sales = np.array([100, 170, 150])
sales.sort()
print(sales)
np.median(sales)
```

```python
sales = np.array([[100, 170, 150],
                  [120, 150, 180],
                  [140, 130, 180]])

np.percentile(sales, 75)

# The 75th percentile means 75% of the values in the sales array are less than or equal to this number, and 25% are greater.
```

```python
np.unique(sales, return_counts=True)
```

```python
np.sqrt(sales)
```

```python
#Standard Deviation 
sales.std()
```

```python
# Position ( index ) of highest and lowest sale
print(np.argmax(sales), np.argmin(sales))
```
