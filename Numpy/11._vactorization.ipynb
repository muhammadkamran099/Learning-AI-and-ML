```python
import numpy as np
```

```python
# Without vectorization: Often slower for large arrays

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])

result = []

for i in range(len(arr1)):
    result.append(arr1[i] * arr2[i])

print(result)
```

```python
# Vectorization: Often faster for large arrays

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])

result = arr1 * arr2

print(result)
```
