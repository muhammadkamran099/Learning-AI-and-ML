```python
import numpy as np
```

```python
ages = np.array([8, 10, 16, 18, 21, 54, 34, 56, 27, 29])

ages[0] = 12
ages
```

```python
ages = np.array([8, 10, 16, 18, 21, 54, 34, 56, 27, 29])

ages[1:4] = [14, 16, 17]
ages
```

```python
scores = np.array([20, 30, 40, 50, 35, 45, 60, 80])
scores[scores > 40] = 50
scores
```

```python
scores = np.array([20, 30, 40, 50, 35, 45, 60, 80])
scores[[3, 6]] = [80, 90]
scores
```
