```python
import numpy as np
```

```python
scores = np.array([45, 34, 60, 43, 75, 48, 80])
score = np.where(scores >= 50, "Pass", "Fail")
score
```

```python
prices = np.array([80, 120, 90, 100, 70, 110])
new_prices = np.where(prices > 100, prices * 0.9, prices)
new_prices
```

```python
charging = np.array([15, 85, 5, 45, 20, 18])
is_charging = np.array(["Falase", "False", "True", "False", "True", "False"])
charge_status = np.where(is_charging == "True", "Charging", np.where(charging > 20, "Charge Now", "OK"))
charge_status
```
