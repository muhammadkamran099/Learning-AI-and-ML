```python
import numpy as np
```

```python
cities = np.array(["Lahore", "Islamabad", "Karachi", "Multan", "Quetta", "Sargodha", "Jehlum", "Peshawar", "Hydrabad", "Sialkot"])
cities = cities.reshape(2, 5)
```

```python
print(cities[1, 3])
print(cities[1, -1])
print(cities[::-1])
print(cities[:,1:4])
print(cities[1, :]) # for a rows
print(cities[:, 1]) # for a columns
```
